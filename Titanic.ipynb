2.-¿Cuántos pasajeros de cada clase aparecen en el archivo? ¿Qué clase tiene más pasajeros registrados?

```python
ds["Clases Del Titanic"]= ds["Pclass"].map({
    1:"Primera clase",
    2:"Segunda clase",
    3:"Tercera clase",
})
```

```python
Numero_Pasajeros_Clase=ds["Clases Del Titanic"].value_counts().sort_index()
tabla_claces=Numero_Pasajeros_Clase.reset_index()
tabla_claces.columns=["Clase Pasajero","Cantidad Total"]
tabla_claces
```

```python
pasajeros_primeraclase= Numero_Pasajeros_Clase["Primera clase"]
pasajeros_segundaclase= Numero_Pasajeros_Clase["Segunda clase"]
pasajeros_terceraclase= Numero_Pasajeros_Clase["Tercera clase"]
```

```python
if (pasajeros_terceraclase>pasajeros_primeraclase and pasajeros_terceraclase>pasajeros_segundaclase):
    print("La tercera clase contiene mas pasajeros con:",pasajeros_terceraclase,"pasajeros")
elif (pasajeros_segundaclase>pasajeros_primeraclase and pasajeros_segundaclase>pasajeros_terceraclase):
    print("La segunda clase contiene mas pasajeros con:",pasajeros_segundaclase,"pasajeros")
else:
    print("La primera clase contiene mas pasajeros con:",pasajeros_primeraclase,"pasajeros")
```

3.-¿Cuántos pasajeros embarcaron en cada puerto y cuántos tienen el puerto sin registrar?

```python
ds["Puertos De Embarque"]= ds["Embarked"].map({
    "C": "Puerto de Cherburgo",
    "Q": "Puerto de Queenstown",
    "S": "Puerto de Southampton",
})
```

```python
ds["Puertos De Embarque"]=ds["Puertos De Embarque"].fillna("Puerto sin registrar")
Conteo_Puertos=ds["Puertos De Embarque"].value_counts().sort_index()
tabla_puertos=Conteo_Puertos.reset_index()
tabla_puertos.columns=["Puerto","Cantidad Total"]
tabla_puertos
```

4.-¿Cuál es la tarifa promedio, mínima y máxima registrada para cada clase?

```python
Tarifas_por_clase=ds.groupby("Clases Del Titanic")["Fare"].agg(["mean","min","max"])
tabla_tarifas=Tarifas_por_clase.reset_index()
tabla_tarifas.columns=["Clase De Pasajeros","Promedio","Minimo","Maximo"]
tabla_tarifas
```

5.-¿Cuántos pasajeros de cada clase embarcaron en cada puerto?

6.-¿Cuántos pasajeros sobrevivieron y cuántos fallecieron en cada clase?

```python
ds["Estado De Pasajeros"]=ds["Survived"].map({
    0: "Fallecio",
    1: "Sobrevivio",
})
```

7.-¿Cuántos pasajeros viajaban con cero, uno o dos o más familiares registrados a bordo?

```python
ds["Total De Familiares"]=ds["SibSp"]+ds["Parch"]
def clasificar_familiares(total):
    if total==0:
        return "Cero familiares"
    elif total==1:
        return "Un familiar"
    else:
        return "Mas de un familiar"
ds["Grupo De Familiares"]=ds["Total De Familiares"].apply(clasificar_familiares)

conteo_familiares=ds["Grupo De Familiares"].value_counts().sort_index()
tabla_conteos=conteo_familiares.reset_index()
tabla_conteos.columns.name=None
tabla_conteos
```

8.-¿Cuántos pasajeros tenían menos de 18 años, cuántos tenían 18 o más y cuántos tenían la edad sin registrar? Desglose el resultado por clase.

```python
def clasificar_edad(edad):
    if lectura.isna(edad):
        return "Sin registrar"
    elif edad<18:
        return "Pasajeros con menos de 18 years"
    else:
        return "Pasajeros con mas de 18 years"
```

```python
ds["Grupo De Edad"]=ds["Age"].apply(clasificar_edad)
tabla_edades= lectura.crosstab(ds["Clases Del Titanic"],ds["Grupo De Edad"])
tabla_edades.columns.name=None
tabla_edades
```

10.-¿En qué clase hay más registros sin cabina? Para cada clase, muestren cuántos registros tienen cabina, cuántos carecen de ella y cuántos hay en total.

```python
def clasificar_cabina(cabina):
    if lectura.isna(cabina):
        return "Sin cabina"
    else:
        return "Con cabina"
ds["Estado_Cabina"]= ds["Cabin"].apply(clasificar_cabina)
```

```python
tabla_registro=lectura.crosstab(ds["Clases Del Titanic"],ds["Estado_Cabina"])
tabla_registro["Total Pasajeros"] = tabla_registro.sum(numeric_only=True,axis=1)
tabla_registro=tabla_registro.reset_index()
tabla_registro.columns.name=None
tabla_registro
```

```python
clase_masregistros_sincabina=tabla_registro.loc[tabla_registro["Sin cabina"].idxmax(),"Clases Del Titanic"]
valor_maximo= tabla_registro["Sin cabina"].max()
print(f"La clase con mas registros 'estados: sin cabina' es: {clase_masregistros_sincabina} con {valor_maximo} pasajeros.")
```
